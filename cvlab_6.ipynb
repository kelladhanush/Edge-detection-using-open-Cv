```python
import cv2
import numpy as np
```

```python
def filteredimage(img, kernel, p, s):
    m = img.shape[0]
    n = img.shape[1]
    km = kernel.shape[0]
    kn = kernel.shape[1]
    rows = int((m - km + 2 * p) / s) + 1
    cols = int((n - kn + 2 * p) / s) + 1
    output = np.ones((rows, cols))
    padding = np.pad(img, p, mode='constant', constant_values=0)
    for i in range(rows):
        for j in range(cols):
            output[i][j] = np.sum(padding[i*s:i*s+km, j*s:j*s+kn] * kernel)
    return output

k = 15
p = 2
s = 2
img = cv2.imread("/content/drive/MyDrive/sanju@2.jpg", 0)
kernel = np.ones((k, k)) / k**2
filtered_output = filteredimage(img, kernel, p, s)
cv2_imshow(filtered_output)
```
